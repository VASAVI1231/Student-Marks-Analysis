Student Marks Analysis

```python
#store student maraks in a list

marks=[78,85,92,67,88]
print("Student Marks:",marks)
```

```python
#Calculate total marks

total=sum(marks)
print("Total Marks:",total)
```

```python
#Calculate averag marks

average=total/len(marks)
print("Average Marks:",average)
```

```python
#Find Highest Marks

highest=max(marks)
print("Heighest Marks:",highest)
```

```python
#Find Lowest Marks

lowest=min(marks)
print("Lowest Marks:",lowest)
```

```python
#Summary
print("Total:",total,
      "Average:",average,
      "Highest:",highest,
      "Lowest:",lowest)
```
